```python
# ============================================================
# STUDENT PASS / FAIL DATASET - COMPLETE 11 STEP ANALYSIS
# ============================================================


# ============================================================
# STEP 1: LOAD AND INSPECT THE DATASET
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

# Upload dataset
uploaded = files.upload()

file_name = list(uploaded.keys())[0]

# Load dataset
df = pd.read_csv(file_name)

print("Dataset loaded successfully!")

print("\nFirst 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())

print("\nDataset description:")
display(df.describe(include="all"))


# ============================================================
# STEP 2: FIND ROWS AND COLUMNS AND CHECK DATA TYPES
# ============================================================

print("\n========== DATASET SHAPE ==========")

rows, columns = df.shape

print("Number of rows:", rows)
print("Number of columns:", columns)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()


# ============================================================
# STEP 3: FIND MISSING VALUES AND DUPLICATE ROWS
# ============================================================

print("\n========== MISSING VALUES ==========")

missing_values = df.isnull().sum()

print(missing_values)

print(
    "\nTotal missing values:",
    df.isnull().sum().sum()
)


print("\n========== DUPLICATE ROWS ==========")

duplicates = df.duplicated().sum()

print(
    "Number of duplicate rows:",
    duplicates
)


# ============================================================
# STEP 4: IDENTIFY INVALID VALUES
# ============================================================

print("\n========== INVALID VALUES ==========")

numeric_columns = df.select_dtypes(
    include=np.number
).columns


# Check all negative numerical values
for col in numeric_columns:

    count = (df[col] < 0).sum()

    if count > 0:

        print(
            f"{col}: {count} negative values"
        )


# Check Study Hours
study_columns = [
    col for col in df.columns
    if (
        "study" in col.lower()
        and "hour" in col.lower()
    )
]

for col in study_columns:

    invalid = (
        df[col] < 0
    ).sum()

    print(
        f"{col}: {invalid} invalid negative study-hour values"
    )


# Check Attendance
attendance_columns = [
    col for col in df.columns
    if "attendance" in col.lower()
]

for col in attendance_columns:

    invalid = (
        (df[col] < 0) |
        (df[col] > 100)
    ).sum()

    print(
        f"{col}: {invalid} invalid attendance values"
    )


# Check Assignments
assignment_columns = [
    col for col in df.columns
    if "assignment" in col.lower()
]

for col in assignment_columns:

    invalid = (
        df[col] < 0
    ).sum()

    print(
        f"{col}: {invalid} invalid assignment values"
    )


# Check Scores
score_columns = [
    col for col in df.columns
    if (
        "score" in col.lower()
        or "mark" in col.lower()
    )
]

for col in score_columns:

    invalid = (
        (df[col] < 0) |
        (df[col] > 100)
    ).sum()

    print(
        f"{col}: {invalid} invalid score values"
    )


# ============================================================
# STEP 5: CHECK OUTLIERS USING BOXPLOTS
# ============================================================

print(
    "\n========== OUTLIER ANALYSIS =========="
)

numeric_columns = df.select_dtypes(
    include=np.number
).columns

for col in numeric_columns:

    plt.figure(figsize=(8, 4))

    sns.boxplot(
        x=df[col]
    )

    plt.title(
        f"Boxplot of {col}"
    )

    plt.xlabel(col)

    plt.show()


# ============================================================
# STEP 6: PLOT STUDY-HOURS DISTRIBUTION
#          AND PASS VS FAIL
# ============================================================

print(
    "\n========== STUDY HOURS DISTRIBUTION =========="
)

if len(study_columns) > 0:

    study_col = study_columns[0]

    plt.figure(figsize=(8, 5))

    plt.hist(
        df[study_col].dropna(),
        bins=20
    )

    plt.title(
        "Study Hours Distribution"
    )

    plt.xlabel(
        study_col
    )

    plt.ylabel(
        "Number of Students"
    )

    plt.show()


# Find Pass/Fail column
pass_fail_columns = [
    col for col in df.columns
    if (
        "pass" in col.lower()
        or "result" in col.lower()
        or "status" in col.lower()
    )
]


if len(pass_fail_columns) > 0:

    pass_fail_col = pass_fail_columns[0]

    print(
        "\nPass/Fail column found:",
        pass_fail_col
    )

    print(
        "\nPass/Fail counts:"
    )

    print(
        df[pass_fail_col].value_counts()
    )

    plt.figure(figsize=(7, 5))

    df[pass_fail_col].value_counts().plot(
        kind="bar"
    )

    plt.title(
        "Pass vs Fail"
    )

    plt.xlabel(
        pass_fail_col
    )

    plt.ylabel(
        "Number of Students"
    )

    plt.xticks(
        rotation=0
    )

    plt.show()


# ============================================================
# STEP 7: ANALYZE STUDY HOURS VS PASS/FAIL
#          AND ATTENDANCE VS PASS/FAIL
# ============================================================

print(
    "\n========== STUDY HOURS VS PASS/FAIL =========="
)

if (
    len(study_columns) > 0
    and len(pass_fail_columns) > 0
):

    study_col = study_columns[0]
    pass_fail_col = pass_fail_columns[0]

    plt.figure(figsize=(8, 5))

    sns.boxplot(
        x=df[pass_fail_col],
        y=df[study_col]
    )

    plt.title(
        "Study Hours vs Pass/Fail"
    )

    plt.xlabel(
        pass_fail_col
    )

    plt.ylabel(
        study_col
    )

    plt.show()


print(
    "\n========== ATTENDANCE VS PASS/FAIL =========="
)

if (
    len(attendance_columns) > 0
    and len(pass_fail_columns) > 0
):

    attendance_col = attendance_columns[0]
    pass_fail_col = pass_fail_columns[0]

    plt.figure(figsize=(8, 5))

    sns.boxplot(
        x=df[pass_fail_col],
        y=df[attendance_col]
    )

    plt.title(
        "Attendance vs Pass/Fail"
    )

    plt.xlabel(
        pass_fail_col
    )

    plt.ylabel(
        attendance_col
    )

    plt.show()


# ============================================================
# STEP 8: CREATE CORRELATION MATRIX / HEATMAP
# ============================================================

print(
    "\n========== CORRELATION MATRIX =========="
)

numeric_df = df.select_dtypes(
    include=np.number
)

correlation_matrix = numeric_df.corr()

print(
    "\nCorrelation Matrix:"
)

display(
    correlation_matrix
)


plt.figure(
    figsize=(10, 7)
)

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title(
    "Student Performance Correlation Heatmap"
)

plt.show()


# ============================================================
# STEP 9: REMOVE DUPLICATES, FIX INVALID VALUES
#          AND HANDLE MISSING VALUES
# ============================================================

print(
    "\n========== DATA CLEANING =========="
)


# Remove duplicate rows
df = df.drop_duplicates()

print(
    "Duplicate rows removed."
)


# Fix Study Hours
for col in study_columns:

    df.loc[
        df[col] < 0,
        col
    ] = np.nan


# Fix Attendance
for col in attendance_columns:

    df.loc[
        (df[col] < 0) |
        (df[col] > 100),
        col
    ] = np.nan


# Fix Assignments
for col in assignment_columns:

    df.loc[
        df[col] < 0,
        col
    ] = np.nan


# Fix Scores
for col in score_columns:

    df.loc[
        (df[col] < 0) |
        (df[col] > 100),
        col
    ] = np.nan


# Handle remaining negative numerical values
numeric_columns = df.select_dtypes(
    include=np.number
).columns

for col in numeric_columns:

    df.loc[
        df[col] < 0,
        col
    ] = np.nan


# Handle missing numerical values
numeric_columns = df.select_dtypes(
    include=np.number
).columns

for col in numeric_columns:

    if df[col].isnull().sum() > 0:

        df[col] = df[col].fillna(
            df[col].median()
        )


# Handle missing categorical values
categorical_columns = df.select_dtypes(
    include="object"
).columns

for col in categorical_columns:

    if df[col].isnull().sum() > 0:

        df[col] = df[col].fillna(
            df[col].mode()[0]
        )


print(
    "\nMissing values after cleaning:"
)

print(
    df.isnull().sum()
)

print(
    "\nDuplicate rows after cleaning:"
)

print(
    df.duplicated().sum()
)

print(
    "\nFinal shape after cleaning:"
)

print(
    df.shape
)


# ============================================================
# STEP 10: CREATE AT LEAST 2 NEW FEATURES
# ============================================================

print(
    "\n========== FEATURE ENGINEERING =========="
)


# FEATURE 1: Study Performance Score

if len(study_columns) > 0:

    study_col = study_columns[0]

    # Create study-hour category
    df["Study_Hour_Category"] = pd.cut(
        df[study_col],
        bins=[
            -1,
            2,
            5,
            8,
            np.inf
        ],
        labels=[
            "Low",
            "Medium",
            "High",
            "Very High"
        ]
    )

    print(
        "Created feature: Study_Hour_Category"
    )


# FEATURE 2: Attendance Category

if len(attendance_columns) > 0:

    attendance_col = attendance_columns[0]

    df["Attendance_Category"] = pd.cut(
        df[attendance_col],
        bins=[
            -1,
            60,
            75,
            85,
            100
        ],
        labels=[
            "Low",
            "Average",
            "Good",
            "Excellent"
        ]
    )

    print(
        "Created feature: Attendance_Category"
    )


# FEATURE 3: Overall Performance
# If score columns are available

if len(score_columns) >= 2:

    df["Overall_Score"] = (
        df[score_columns].mean(axis=1)
    )

    print(
        "Created feature: Overall_Score"
    )


print(
    "\nDataset after feature engineering:"
)

display(
    df.head()
)


# ============================================================
# STEP 11: SEPARATE X AND y
# ============================================================

print(
    "\n========== X AND y SEPARATION =========="
)


if len(pass_fail_columns) > 0:

    target_column = pass_fail_columns[0]

    # y = target variable
    y = df[target_column]

    # X = input features
    X = df.drop(
        columns=[target_column]
    )

    print(
        "Target column (y):",
        target_column
    )

    print(
        "\nX shape:",
        X.shape
    )

    print(
        "y shape:",
        y.shape
    )

    print("\nX:")
    display(
        X.head()
    )

    print("\ny:")
    display(
        y.head()
    )

else:

    print(
        "Pass/Fail target column not found."
    )


# ============================================================
# FINAL SUMMARY
# ============================================================

print(
    "\n========================================"
)

print(
    "       STUDENT PASS / FAIL ANALYSIS"
)

print(
    "                COMPLETED"
)

print(
    "========================================"
)

print(
    "\nFinal dataset shape:"
)

print(
    df.shape
)

print(
    "\nFinal columns:"
)

print(
    df.columns.tolist()
)

print(
    "\nMissing values:"
)

print(
    df.isnull().sum()
)

print(
    "\nDuplicate rows:"
)

print(
    df.duplicated().sum()
)

print(
    "\nFinal dataset:"
)

display(
    df.head()
)
```
